# Neural Network Linear Algebra Project
Applied Linear Algebra final project implemented in Python/Jupyter Notebook.

This notebook verifies a complete forward pass and backpropagation calculation for a small 3 → 4 → 1 neural network using one Iris dataset observation.


```python
import numpy as np
from sklearn.datasets import load_iris

seed = 20
np.random.seed(seed)

n_input = 3
n_hidden = 4
n_output = 1
```

## Weight Initialization
The network uses a 4×3 first-layer weight matrix and a 1×4 second-layer weight matrix. Biases are initialized to zero.


```python
W1 = np.random.randn(n_hidden, n_input) * 0.5
b1 = np.zeros(n_hidden)
W2 = np.random.randn(n_output, n_hidden) * 0.5
b2 = np.zeros(n_output)

print("W1:\n", np.round(W1, 4))
print("b1:", b1)
print("W2:\n", np.round(W2, 4))
print("b2:", b2)
```

## Iris Regression Example
Inputs are sepal length, sepal width, and petal width. The target is petal length. The first Iris observation is used for manual and computational verification.


```python
data = load_iris()
X_full = data.data[:, [0, 1, 3]]
y_full = data.data[:, 2]

x = X_full[0]
y = y_full[0]

print("x =", x)
print("target y =", y)
```

## Forward Pass
Hidden pre-activation: a1 = W1x + b1
Hidden activation: z1 = tanh(a1)
Output: a2 = W2z1 + b2
Loss: 1/2(ŷ-y)^2


```python
a1 = W1 @ x + b1
z1 = np.tanh(a1)
a2 = W2 @ z1 + b2
y_hat = a2[0]
loss = 0.5 * (y_hat - y) ** 2

print("Hidden pre-activations:", np.round(a1, 4))
print("Hidden activations:", np.round(z1, 4))
print("Prediction:", round(y_hat, 4))
print("Target:", y)
print("Loss:", round(loss, 4))
```

## Backpropagation
Gradients are derived with the chain rule. The tanh derivative is 1-z².


```python
# Output-layer gradient
dE_da2 = y_hat - y
delta2 = dE_da2
dE_dW2 = delta2 * z1.reshape(1, -1)
dE_db2 = delta2

# Propagate error to hidden layer
delta1_z = W2.T.flatten() * delta2
h_prime_a1 = 1 - z1**2
delta1_a = delta1_z * h_prime_a1

# First-layer gradients
dE_dW1 = np.outer(delta1_a, x)
dE_db1 = delta1_a.copy()

print("dE/da2:", round(dE_da2, 4))
print("dE/dW2:\n", np.round(dE_dW2, 4))
print("dE/db2:", round(dE_db2, 4))
print("Hidden error signal:", np.round(delta1_z, 4))
print("tanh derivative:", np.round(h_prime_a1, 4))
print("dE/da1:", np.round(delta1_a, 4))
print("dE/dW1:\n", np.round(dE_dW1, 4))
print("dE/db1:", np.round(dE_db1, 4))
```

## Chain-Rule Verification
The derivative for a specific first-layer weight is also checked as an explicit product of chain-rule factors.


```python
factor1 = dE_da2
factor2 = W2[0, 0]
factor3 = h_prime_a1[0]
factor4 = x[1]
chain_product = factor1 * factor2 * factor3 * factor4

print("Explicit chain-rule product:", round(chain_product, 4))
print("Gradient-matrix value:", round(dE_dW1[0, 1], 4))
print("Match:", np.isclose(chain_product, dE_dW1[0, 1]))
```
